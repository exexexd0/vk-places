Цикл выбора точек интереса. Перед запуском поднимите Postgres, заполните `.env` и выполните `python -m rag rebuild`. Ключ читается только из окружения.

```python
from rag.graph import build_route_graph, initial_state, route_view
from rag.jev import JevSelector
from rag.planner import LlmPlanner
from rag.settings import load_settings
from rag.store import available_categories, open_store

settings = load_settings()
store = await open_store(settings)
categories = available_categories(settings)
app = build_route_graph(LlmPlanner(settings), store, JevSelector(settings))
state = await app.ainvoke(
    initial_state(
        "Хочу поесть и зайти в торговый центр в Иваново",
        settings.city,
        categories,
    )
)
route_view(state)
```
